### Handling imbalance dataset
1. up sampling ------> Increase minority class samples.
2. down sampling ----> Decrease majority class samples.

```python
import numpy as np
import pandas as pd

## set the random seed for the reproducibility
np.random.seed(123)

n_sample = 1000

## create the dataframe with two class
n_class_0_ratio = 0.9
n_class_0 = int(n_sample * n_class_0_ratio)
n_class_1 = n_sample - n_class_0
```

```python
n_class_0,n_class_1
```

```python
## Creating the Dataframe with imbalance Dataset

class_0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_0), ## loc = mean, scale=standard divation
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'target' : [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'target' : [1] * n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df['target'].value_counts()
```

```python
df.sample(5)
```

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

### UP Sampling

```python
## up sampling -------> increase the minority class for the data imbalanced
from sklearn.utils import resample
```

```python
df_minority_upsampled = resample(df_minority, replace=True, ## same row can be selected multiple times
                                 n_samples=len(df_majority), random_state=42)
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head(5)
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```

```python
df_upsampled.shape
```
