### SMOTE(synthetic minority over-sampling technique)
#### It is a technique used to handle imbalanced datasets by creating new synthetic (artificial) samples for the minority class instead of simply copying existing samples.

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000, n_features=2, n_redundant=0, n_clusters_per_class=1, 
                   weights=[0.90], random_state=12)
```

```python
## creating the dataframe
import pandas as pd
df1 = pd.DataFrame(X,columns=['f1','f2'])
df2 = pd.DataFrame(y,columns=['target'])
```

```python
y.shape
```

```python
X.shape
```

```python
final_df = pd.concat([df1,df2],axis=1)
final_df.shape
```

```python
final_df.sample(5)
```

```python
final_df['target'].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(final_df['f1'],final_df['f2'],c=final_df['target'])
plt.show()
```

```python
import imblearn
print(imblearn.__version__)
```

```python
from imblearn.over_sampling import SMOTE
```

```python
## transform the data
over_sampling = SMOTE()
X,y = over_sampling.fit_resample(final_df[['f1','f2']],final_df['target'])
```

```python
X.shape
```

```python
y.shape
```

```python
## concat the dataframe X,y
```

```python
print(len(y[y==0]))
print(len(y[y==1]))
```

```python
df1 = pd.DataFrame(X,columns=['f1','f2'])
df2 = pd.DataFrame(y,columns=['target'])
```

```python
df_smote_sampled = pd.concat([df1,df2],axis=1)
```

```python
plt.scatter(df_smote_sampled['f1'],df_smote_sampled['f2'],
            c=df_smote_sampled['target'])
```
